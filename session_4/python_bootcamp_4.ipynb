## Probability 101

As mentioned earlier, probability is the elegant science of randomness. Granted, randomness is random, but it still follows patterns. Sometimes it is centered around a certain value, other times is evenly distributed along a range of values. Sometimes it is only integers (i.e. throwing a 2-sided coin or 6-sided dice), other times it can be any decimal number (i.e. the exact height of someone or the temperature outside). Probability neatly organizes all these features and studies their implications and connections. 

You might already be familiar with some version of probability. Something like: given a set of 52 playing cards, what is the probability of picking a heart? In which case, you would instinctively divide the number of hearts in deck (13), by the total number of cards in the deck (52), which would correctly yield $1/4$. There are infinite variations of this problem, where the set of all possible outcomes (aka the sample space, $S$), just like a deck of cards, is finite; where each element is equally likely to be sampled; and where we are interested in the probability of selecting a particular subset of samples (usually denoted by a letter like $A$, $B$, $C$, and so on). In this case, the probability of picking a subset of the sample space is simply the size of the subset (denoted $|A|$) divided by the size of the sample space (denoted $|S|$):
$$P(A) = \frac{|A|}{|S|}$$

For example, say we have one dice and want to calculate the probability that we roll an even number. $S = \{1,2,3,4,5,6\}$, $A = \{2,4,6\}$, and thus, $P(A) = |A|/|S| = 3/6 = 1/2$. Of course, this problem can be a lot more complicated than simple counting on your fingers. The field of mathematics that deals with counting in complex ways is called **combinatorics**. Some useful combinatorics tricks you might have seen before include:

| The name | The math | The meaning |  
| --- | --- | --- |
| Factorial | $n!$ | The number of permutations of $n$ distinct items (the number of ways they can be reordered) |
| Binomial coefficient | $\binom{n}{k} = \frac{n!}{k!(n-k)!}$ | Pronounced "$n$ choose $k$"; the number of ways you can choose $k$ items from $n$ distinct items **without** replacement |
| | $n^k$ | The number of ways you can choose $k$ items from $n$ distinct items **with** replacement
